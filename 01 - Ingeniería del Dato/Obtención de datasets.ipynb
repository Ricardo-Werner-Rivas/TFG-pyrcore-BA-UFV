## Bolsa china

### Obtención de datos

```python
# Clonar repositorio "Multivariate_Stock_Time_Series_Dataset" de "shaocongWu"
!git clone https://github.com/shaocongWu/Multivariate_Stock_Time_Series_Dataset.git
```

```python
# Crear el DataFrame con todos los datos por concatenación (pd.concat)
df=pd.concat(
    # Lista de DataFrames
    [
        # DataFrame con los datos del fichero CSV
        pd.read_csv(
            f"Multivariate_Stock_Time_Series_Dataset\\CSV\\{fichero}", # Ruta del fichero
            index_col=0 # La primera columna del fichero se usará como índice
        )
        # Bucle for comprimido
        for fichero in listdir("Multivariate_Stock_Time_Series_Dataset\\CSV")
    ],
    ignore_index=True # La columna de índices se rehará para el DataFrame concatenado
)
# Mostrar la información del DataFrame
df.info()
```

```python
# Eliminar repositorio clonado
!rmdir /s /q Multivariate_Stock_Time_Series_Dataset
```

### Base de datos

```python
# Almacenar DataFrame en una tabla llamada "ACCIONES"
df.to_sql("ACCIONES",conn,index=False)
# Guardar los cambios hechos en la base de datos
conn.commit()
# Interrumpir la conexión con la base de datos
conn.close()
```
