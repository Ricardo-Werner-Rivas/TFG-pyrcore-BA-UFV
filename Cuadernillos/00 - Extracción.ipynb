### Obtención de datos

```python
# Obtener DataFrame desde fichero Excel descargado localmente
df=pd.read_excel("..\\datasets\\data_shale_gas_010822.xlsx",header=1)
# Previsualizar datos
df.head()
```

### Base de datos

```python
# Conectar a la base de datos creando un archivo DB
conn=sqlite3.connect("..\\datasets\\Base de datos TFG.db")
# Almacenar DataFrame
df.to_sql("Esquisto",conn,index=False)
# Guardar cambios en la base de datos
conn.commit()
# Desconectar de la base de datos
conn.close()
```
